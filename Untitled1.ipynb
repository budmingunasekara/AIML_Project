```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer

sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)

df = pd.read_csv('heart_disease_combined.csv')
print('Shape:', df.shape)
df.head()
```

```python
df.info()
print('\nMissing values per column:')
print(df.isnull().sum().sort_values(ascending=False))
print('\nTarget distribution:')
print(df['target'].value_counts().sort_index())
print('\nSource distribution:')
print(df['source'].value_counts())
```

```python
df['target_binary'] = (df['target'] > 0).astype(int)
df[['target', 'target_binary']].value_counts().sort_index()
```

```python
numeric_cols = ['age', 'trestbps', 'chol', 'thalach', 'oldpeak']
categorical_num_cols = ['sex', 'cp', 'fbs', 'restecg', 'exang', 'slope', 'ca', 'thal']

df_clean = df.copy()

num_imputer = SimpleImputer(strategy='median')
df_clean[numeric_cols] = num_imputer.fit_transform(df_clean[numeric_cols])

cat_imputer = SimpleImputer(strategy='most_frequent')
df_clean[categorical_num_cols] = cat_imputer.fit_transform(df_clean[categorical_num_cols])

print('Remaining missing values:')
print(df_clean.isnull().sum().sum())
```

```python
# Age groups
df_clean['age_group'] = pd.cut(
    df_clean['age'], bins=[0, 40, 50, 60, 70, 100],
    labels=['<40', '40-50', '50-60', '60-70', '70+']
)

# Cholesterol risk category (standard clinical thresholds)
df_clean['chol_category'] = pd.cut(
    df_clean['chol'], bins=[-1, 200, 240, 1000],
    labels=['normal', 'borderline_high', 'high']
)

# Resting blood pressure category
df_clean['bp_category'] = pd.cut(
    df_clean['trestbps'], bins=[-1, 120, 140, 1000],
    labels=['normal', 'elevated', 'high']
)

# Max heart rate reserve proxy: how close thalach is to the age-predicted max (220 - age)
df_clean['hr_reserve_pct'] = df_clean['thalach'] / (220 - df_clean['age'])

# Combined risk flag: exercise-induced angina AND ST depression present
df_clean['exercise_stress_flag'] = ((df_clean['exang'] == 1) & (df_clean['oldpeak'] > 1)).astype(int)

df_clean[['age', 'age_group', 'chol', 'chol_category', 'trestbps', 'bp_category',
          'hr_reserve_pct', 'exercise_stress_flag']].head()
```

```python
nominal_cols = ['cp', 'restecg', 'slope', 'thal', 'age_group', 'chol_category', 'bp_category']

df_encoded = pd.get_dummies(df_clean, columns=nominal_cols, prefix=nominal_cols)

# One-hot encode source too (which study the record came from)
df_encoded = pd.get_dummies(df_encoded, columns=['source'], prefix='source')

print('Shape after encoding:', df_encoded.shape)
df_encoded.head()
```

```python
scale_cols = ['age', 'trestbps', 'chol', 'thalach', 'oldpeak', 'hr_reserve_pct']

scaler = StandardScaler()
df_encoded[scale_cols] = scaler.fit_transform(df_encoded[scale_cols])

df_encoded[scale_cols].describe()
```

```python
import os, json, time, warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib

from sklearn.base import clone
from sklearn.model_selection import train_test_split, StratifiedKFold, GridSearchCV, cross_validate
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score,
                             roc_auc_score, confusion_matrix, ConfusionMatrixDisplay,
                             RocCurveDisplay, classification_report)
from sklearn.inspection import permutation_importance

warnings.filterwarnings("ignore", category=FutureWarning)
warnings.filterwarnings("ignore", category=UserWarning)
sns.set_style("whitegrid")

# ---------------- configuration ----------------
RANDOM_STATE = 42
USE_SOURCE   = False     # False = do NOT use hospital/source as a predictor (see report: ethics/shortcut learning)
DATA_FILE    = "/content/heart_disease_combined (1).csv"   # file produced by the group's cleaning steps
RESULTS_DIR  = "results/models"
os.makedirs(RESULTS_DIR, exist_ok=True)

# ---------------- data loading (stateless steps only) ----------------
def load_heart_data(path=None):
    candidates = [path] if path else [DATA_FILE, "/content/" + DATA_FILE,
                                      "/content/sample_data/" + DATA_FILE]
    found = next((p for p in candidates if p and os.path.exists(p)), None)
    if found is None:
        raise FileNotFoundError(f"Upload {DATA_FILE} next to this notebook. Looked in: {candidates}")
    df = pd.read_csv(found)
    n0 = len(df)
    df = df.drop_duplicates().reset_index(drop=True)
    print(f"Loaded {found}: {n0} rows -> {len(df)} after dropping exact duplicates")

    # Physiologically impossible zeros are really missing values (imputed later, inside the pipeline)
    for c in ["chol", "trestbps"]:
        df.loc[df[c] == 0, c] = np.nan

    # Binary target: 0 = no disease, 1-4 = disease present
    df["target_binary"] = (df["target"] > 0).astype(int)

    # Row-wise engineered features (no statistics learned from other rows -> no leakage)
    df["hr_reserve_pct"] = df["thalach"] / (220 - df["age"])
    df["exercise_stress_flag"] = ((df["exang"] == 1) & (df["oldpeak"] > 1)).astype(int)
    return df

NUM = ["age", "trestbps", "chol", "thalach", "oldpeak", "hr_reserve_pct", "ca"]
NOM = ["cp", "restecg", "slope", "thal"] + (["source"] if USE_SOURCE else [])
BIN = ["sex", "fbs", "exang", "exercise_stress_flag"]
FEATURES = NUM + NOM + BIN

df = load_heart_data()
for c in NOM:   # nominal codes -> strings so 'missing' can become its own category
    if c != "source":
        df[c] = df[c].map(lambda v: v if pd.isna(v) else str(v))

X = df[FEATURES]
y = df["target_binary"]

# ---------------- preprocessing (unfitted; fitted inside each Pipeline) ----------------
def make_preprocessor():
    num = Pipeline([("imp", SimpleImputer(strategy="median", add_indicator=True)),
                    ("sc",  StandardScaler())])
    nom = Pipeline([("imp", SimpleImputer(strategy="constant", fill_value="missing")),
                    ("ohe", OneHotEncoder(handle_unknown="ignore"))])
    bin_ = Pipeline([("imp", SimpleImputer(strategy="most_frequent"))])
    return ColumnTransformer([("num", num, NUM), ("nom", nom, NOM), ("bin", bin_, BIN)])

# ---------------- split + validation scheme ----------------
strata = df["target_binary"].astype(str) + "_" + df["source"]
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, stratify=strata, random_state=RANDOM_STATE)
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
SCORING = {"accuracy": "accuracy", "precision": "precision",
           "recall": "recall", "f1": "f1", "roc_auc": "roc_auc"}

print(f"Train: {X_train.shape}  Test: {X_test.shape}")
print("Class balance  train:", y_train.value_counts(normalize=True).round(3).to_dict(),
      " test:", y_test.value_counts(normalize=True).round(3).to_dict())

# ---------------- helper functions ----------------
def _scores(model, X_):
    return model.predict_proba(X_)[:, 1] if hasattr(model, "predict_proba") else model.decision_function(X_)

def _metrics(y_true, y_pred, y_score):
    return {"accuracy":  accuracy_score(y_true, y_pred),
            "precision": precision_score(y_true, y_pred, zero_division=0),
            "recall":    recall_score(y_true, y_pred),
            "f1":        f1_score(y_true, y_pred),
            "roc_auc":   roc_auc_score(y_true, y_score)}

def baseline_eval(pipe, label="Baseline (default hyper-parameters)"):
    """5-fold CV on the training set + hold-out test score of the untuned model."""
    cvres = cross_validate(pipe, X_train, y_train, cv=cv, scoring=SCORING, n_jobs=-1)
    m = clone(pipe).fit(X_train, y_train)
    test = _metrics(y_test, m.predict(X_test), _scores(m, X_test))
    out = {"cv_mean": {k: float(cvres[f"test_{k}"].mean()) for k in SCORING},
           "cv_std":  {k: float(cvres[f"test_{k}"].std())  for k in SCORING},
           "test": {k: float(v) for k, v in test.items()}}
    print(label)
    print(pd.DataFrame({"CV mean": out["cv_mean"], "CV std": out["cv_std"], "Test": out["test"]}).round(3))
    return out

def run_grid_search(pipe, param_grid):
    gs = GridSearchCV(pipe, param_grid, scoring=SCORING, refit="f1", cv=cv,
                      n_jobs=-1, return_train_score=True)
    t0 = time.time(); gs.fit(X_train, y_train)
    n_cand = len(gs.cv_results_["params"])
    print(f"Candidates: {n_cand} | folds: {cv.get_n_splits()} | model fits: {n_cand * cv.get_n_splits()} "
          f"(+1 final refit) | time: {time.time() - t0:.1f}s")
    print("Best params :", gs.best_params_)
    print(f"Best CV F1  : {gs.best_score_:.3f}")
    return gs

def top_configs(gs, n=5):
    r = pd.DataFrame(gs.cv_results_).sort_values("rank_test_f1").head(n)
    cols = ["params", "mean_test_f1", "std_test_f1", "mean_test_recall", "mean_test_precision",
            "mean_test_accuracy", "mean_test_roc_auc", "mean_train_f1"]
    return r[cols].rename(columns=lambda c: c.replace("mean_test_", "cv_").replace("std_test_", "cv_std_")
                          .replace("mean_train_", "train_")).round(3).reset_index(drop=True)

_ABSENT = "__absent__"
def plot_param_effect(gs, param, ax=None, title=None):
    """Best CV F1 (and the matching train F1) for each value of one hyper-parameter.
       A widening gap between the two curves = over-fitting."""
    r = pd.DataFrame(gs.cv_results_)
    vals = r["params"].apply(lambda p: p.get(param, _ABSENT))
    r = r[vals.astype(str) != _ABSENT].copy()
    r["_v"] = vals[r.index].astype(str)
    idx = r.groupby("_v", sort=False)["mean_test_f1"].idxmax()
    best = r.loc[idx.values]
    ax = ax or plt.subplots(figsize=(7, 4))[1]
    ax.plot(best["_v"], best["mean_train_f1"], "o--", label="Train F1")
    ax.plot(best["_v"], best["mean_test_f1"], "o-", label="CV (validation) F1")
    ax.fill_between(range(len(best)), best["mean_test_f1"] - best["std_test_f1"],
                    best["mean_test_f1"] + best["std_test_f1"], alpha=0.15)
    ax.set_xlabel(param.split("__")[-1]); ax.set_ylabel("F1")
    ax.set_title(title or f"Effect of {param.split('__')[-1]}"); ax.legend()
    plt.setp(ax.get_xticklabels(), rotation=45, ha="right")
    return ax

def compare_varieties(gs, param):
    """For every value of `param` take its best CV configuration, refit it on the training set
       and report CV + hold-out test metrics -> a like-for-like comparison of model varieties."""
    r = pd.DataFrame(gs.cv_results_)
    vals = r["params"].apply(lambda p: str(p.get(param, _ABSENT)))
    rows = []
    for v in pd.unique(vals[vals != _ABSENT]):
        sub = r[vals == v]
        b = sub.loc[sub["mean_test_f1"].idxmax()]
        m = clone(gs.estimator).set_params(**b["params"]).fit(X_train, y_train)
        t = _metrics(y_test, m.predict(X_test), _scores(m, X_test))
        rows.append({param.split("__")[-1]: v, "cv_f1": b["mean_test_f1"], "cv_recall": b["mean_test_recall"],
                     "test_accuracy": t["accuracy"], "test_precision": t["precision"],
                     "test_recall": t["recall"], "test_f1": t["f1"], "test_auc": t["roc_auc"],
                     "best_params": b["params"]})
    return pd.DataFrame(rows).round(3)

def evaluate_search(name, gs, baseline=None):
    """Full evaluation of the tuned model: CV, train vs test, confusion matrix, ROC. Saves everything."""
    best = gs.best_estimator_; i = gs.best_index_; res = gs.cv_results_
    cv_mean = {k: float(res[f"mean_test_{k}"][i]) for k in SCORING}
    cv_std  = {k: float(res[f"std_test_{k}"][i])  for k in SCORING}
    y_pred = best.predict(X_test); y_score = _scores(best, X_test)
    test  = {k: float(v) for k, v in _metrics(y_test, y_pred, y_score).items()}
    train = {k: float(v) for k, v in _metrics(y_train, best.predict(X_train), _scores(best, X_train)).items()}

    print(f"=== {name}: tuned model ===")
    print(pd.DataFrame({"CV mean (5-fold)": cv_mean, "CV std": cv_std, "Train": train, "Test (hold-out)": test}).round(3))
    print(f"\nTrain-Test F1 gap: {train['f1'] - test['f1']:+.3f}  (large positive gap => over-fitting)\n")
    print(classification_report(y_test, y_pred, target_names=["No disease", "Disease"]))

    cm = confusion_matrix(y_test, y_pred); tn, fp, fn, tp = cm.ravel()
    fig, ax = plt.subplots(1, 2, figsize=(11, 4.5))
    ConfusionMatrixDisplay(cm, display_labels=["No disease", "Disease"]).plot(ax=ax[0], cmap="Blues", colorbar=False)
    ax[0].set_title(f"{name} - confusion matrix (test)")
    RocCurveDisplay.from_predictions(y_test, y_score, ax=ax[1], name=f"{name} (AUC={test['roc_auc']:.3f})")
    ax[1].plot([0, 1], [0, 1], "k--", alpha=.4); ax[1].set_title(f"{name} - ROC curve (test)")
    plt.tight_layout(); plt.savefig(f"{RESULTS_DIR}/{name}_cm_roc.png", dpi=150); plt.show()
    print(f"False negatives (missed patients): {fn} | False positives (false alarms): {fp}")

    def _py(v): return str(v) if isinstance(v, tuple) else (v.item() if hasattr(v, "item") else v)
    out = {"model": name,
           "best_params": {k: _py(v) for k, v in gs.best_params_.items()},
           "n_candidates": len(res["params"]), "n_fits": len(res["params"]) * cv.get_n_splits(),
           "cv_mean": cv_mean, "cv_std": cv_std, "train": train, "test": test,
           "confusion": {"tn": int(tn), "fp": int(fp), "fn": int(fn), "tp": int(tp)},
           "baseline": baseline}
    with open(f"{RESULTS_DIR}/{name}_metrics.json", "w") as f: json.dump(out, f, indent=2)
    pd.DataFrame({"y_true": y_test.values, "y_pred": y_pred, "y_score": y_score}).to_csv(
        f"{RESULTS_DIR}/{name}_test_predictions.csv", index=False)
    pd.DataFrame(res).drop(columns=["params"]).to_csv(f"{RESULTS_DIR}/{name}_cv_results.csv", index=False)
    joblib.dump(best, f"{RESULTS_DIR}/{name}_best_model.joblib")
    print(f"Saved metrics, predictions, CV table, plot and model to {RESULTS_DIR}/")
    return out

def plot_permutation_importance(model, name, n_repeats=20, top=12):
    """Model-agnostic importance on the ORIGINAL features: drop in test F1 when a feature is shuffled."""
    pi = permutation_importance(model, X_test, y_test, scoring="f1", n_repeats=n_repeats,
                                random_state=RANDOM_STATE, n_jobs=-1)
    s = pd.Series(pi.importances_mean, index=X_test.columns).sort_values().tail(top)
    err = pd.Series(pi.importances_std, index=X_test.columns)[s.index]
    ax = s.plot.barh(xerr=err, figsize=(7, 5), color="steelblue")
    ax.set_title(f"{name} - permutation importance (drop in test F1)"); ax.set_xlabel("Mean F1 decrease")
    plt.tight_layout(); plt.savefig(f"{RESULTS_DIR}/{name}_perm_importance.png", dpi=150); plt.show()
    return s.sort_values(ascending=False)
```
