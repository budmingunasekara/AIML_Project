```python
!pip install streamlit
import streamlit as st
import pickle
import numpy as np
import pandas as pd
import os
from pathlib import Path
import pickle, warnings
```

```python
warnings.filterwarnings("ignore")
st.set_page_config(page_title="Heart Disease Risk Predictor", page_icon="❤️", layout="centered")
```

```python
@st.cache_resource
def load_model():
    files = sorted(Path(__file__).parent.glob("heart-disease-model*.pkl"))
    if not files:
        raise FileNotFoundError("Put your model file (heart-disease-model.pkl) in the same folder as app.py")
    with open(files[0], "rb") as f:
        return pickle.load(f)

try:
    model = load_model()
except Exception as e:
    st.error(f"Could not load the model: {e}")
    st.info("The model was saved with scikit-learn 1.6.1. Install the same version:  "
            "`pip install scikit-learn==1.6.1`")
    st.stop()
```

```python

SEX      = {"Female": 0, "Male": 1}
CP       = {"Typical angina": 1, "Atypical angina": 2, "Non-anginal pain": 3, "Asymptomatic": 4}
RESTECG  = {"Normal": 0, "ST-T wave abnormality": 1, "Left ventricular hypertrophy": 2}
SLOPE    = {"Unknown": np.nan, "Upsloping": 1, "Flat": 2, "Downsloping": 3}
CA       = {"Unknown": np.nan, "0 vessels": 0, "1 vessel": 1, "2 vessels": 2, "3 vessels": 3}
THAL     = {"Unknown": np.nan, "Normal": 3, "Fixed defect": 6, "Reversible defect": 7}
```

```python
st.title("❤️ Heart Disease Risk Predictor")
st.caption("IT2011 AI & ML project · Random Forest model trained on the combined UCI heart-disease data "
           "(Cleveland, Hungarian, Switzerland, VA) and this project from Y2-S1-MLB-B07G1-07")

```

```python
with st.form("patient_form"):
    st.subheader("Patient details")
    c1, c2, c3 = st.columns(3)
    age  = c1.number_input("Age (years)", 18, 100, 54)
    sex  = c2.selectbox("Sex", list(SEX))
    fbs  = c3.selectbox("Fasting blood sugar > 120 mg/dl", ["No", "Yes"])

    st.subheader("Clinical measurements")
    c1, c2, c3 = st.columns(3)
    trestbps = c1.number_input("Resting blood pressure (mm Hg)", 70, 250, 130)
    chol     = c2.number_input("Cholesterol (mg/dl)", 80, 700, 240)
    chol_unknown = c2.checkbox("Cholesterol not available")
    restecg  = c3.selectbox("Resting ECG", list(RESTECG))

    st.subheader("Symptoms & exercise test")
    c1, c2 = st.columns(2)
    cp      = c1.selectbox("Chest pain type", list(CP), index=3)
    exang   = c2.selectbox("Exercise-induced angina", ["No", "Yes"])
    c1, c2, c3 = st.columns(3)
    thalach = c1.number_input("Max heart rate achieved (bpm)", 60, 230, 143)
    oldpeak = c2.number_input("ST depression (oldpeak)", 0.0, 7.0, 0.5, step=0.1)
    slope   = c3.selectbox("Slope of peak exercise ST segment", list(SLOPE))

    st.subheader("Imaging / thallium test")
    c1, c2 = st.columns(2)
    ca   = c1.selectbox("Major vessels coloured by fluoroscopy", list(CA))
    thal = c2.selectbox("Thallium stress test result", list(THAL))

    submitted = st.form_submit_button("Predict", type="primary", width="stretch")
```

```python
if submitted:
    row = pd.DataFrame([{
        "age": age, "sex": SEX[sex], "cp": CP[cp],
        "trestbps": trestbps, "chol": np.nan if chol_unknown else chol,
        "fbs": 1 if fbs == "Yes" else 0, "restecg": RESTECG[restecg],
        "thalach": thalach, "exang": 1 if exang == "Yes" else 0, "oldpeak": oldpeak,
        "slope": SLOPE[slope], "ca": CA[ca], "thal": THAL[thal],
    }]).astype(float)[list(model.feature_names_in_)]      # same column order as training

    prob = float(model.predict_proba(row)[0, 1])
    n_missing = int(row.isna().sum(axis=1).iloc[0])

    st.divider()
    st.subheader("Result")
    left, right = st.columns([1, 2])
    left.metric("Probability of heart disease", f"{prob:.0%}")
    with right:
        st.progress(prob)
        if prob >= 0.5:
            st.error("**Model prediction: heart disease likely.** Medical follow-up is recommended.")
        else:
            st.success("**Model prediction: heart disease unlikely.** This does not rule it out.")
    if n_missing:
        st.info(f"{n_missing} input(s) were marked unknown and filled with typical values by the model "
                "(median / most frequent) – the prediction is less reliable.")

    with st.expander("Values sent to the model"):
        st.dataframe(row.T.rename(columns={0: "value"}), width="stretch")
```

```python
if 'model' in globals():
    with st.expander("About the model & limitations"):
        clf = model[-1]
        names = model[0].get_feature_names_out()
        imp = pd.Series(clf.feature_importances_, index=[n.split("__")[-1] for n in names]).sort_values(ascending=False).head(10)
        st.write(f"**Algorithm:** Random Forest ({clf.n_estimators} trees, max depth {clf.max_depth}, "
                 f"class_weight = {clf.class_weight}). Missing values are imputed inside the model pipeline.")
        st.write("**Top 10 features by importance:**")
        st.bar_chart(imp)
        st.markdown(
            "- Trained on ~900 patients from four hospitals; mostly male and older, so it may be less accurate for women and younger people.\n"
            "- A *false negative* (a sick patient predicted healthy) is the most serious error – never use this to decide against seeing a doctor.\n"
            "- Prediction is a probability from past data, not a diagnosis.")
else:
    st.warning("Model is not loaded. Please ensure the model file exists and cell Stq6vLVpGQ6J runs successfully.")
```
